```python
import zipfile
import pandas as pd
import io


zip_file_path = "C:/Users/pc/Desktop/DEPI/final project/logistics-operations-database.zip"

logistics_tables = {}


with zipfile.ZipFile(zip_file_path, 'r') as archive:
   
    print("الملفات الموجودة جوه الـ zip:")
    print(archive.namelist())
    print("-" * 50)
    
    for file_path in archive.namelist():
        
        if file_path.endswith('.csv'):
            
            file_name = file_path.split('/')[-1].replace('.csv', '')
            
            with archive.open(file_path) as file:
                logistics_tables[file_name] = pd.read_csv(file)
                print(f"تم تحميل جدول [{file_name}] بنجاح! الأبعاد: {logistics_tables[file_name].shape}")
```

```python
print("--- Skewness Calculation ---")
print(customers[['credit_terms_days', 'annual_revenue_potential']].skew())

print("\n--- Outliers Detection via IQR ---")
for col in ['credit_terms_days', 'annual_revenue_potential']:
    q1 = customers[col].quantile(0.25)
    q3 = customers[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr
    outliers = customers[(customers[col] < lower_bound) | (customers[col] > upper_bound)]
    print(f"{col} outliers count: {len(outliers)}")

print("\n--- Categorical Columns Unique Values ---")
print(customers['customer_type'].value_counts())
print("-" * 30)
print(customers['account_status'].value_counts())
print("-" * 30)
print(customers['primary_freight_type'].value_counts())

customers['contract_start_date'] = pd.to_datetime(customers['contract_start_date'])

print("\n--- Final Data Types Verification ---")
print(customers.dtypes)
```

```python
import os

text_columns = ['customer_id', 'customer_name', 'customer_type', 'primary_freight_type', 'account_status']
for col in text_columns:
    customers[col] = customers[col].astype(str).str.strip()

save_path = r"C:\Users\pc\Desktop\DEPI\final project\data cleaned"
os.makedirs(save_path, exist_ok=True)

final_file_path = os.path.join(save_path, "customers.csv")
customers.to_csv(final_file_path, index=False)

print("Text columns stripped successfully")
print("Data types verified")
print("File successfully saved to:", final_file_path)
```
